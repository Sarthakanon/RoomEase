# 02 - Feature Engineering V2

This notebook explains the model-ready features created by `feature_engineering_v2.py`.

Feature folder: `features_v2_nepal/`

## Feature Safety

The important rule is avoiding leakage. Forecast and risk features use prior-period rolling values where future prediction is required.

Feature groups:

- monthly features for next-month spending forecast and budget risk
- daily features for anomaly detection
- shared expense features for settlement delay risk

```python
from pathlib import Path
import json
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks_v2' else Path.cwd()
FEATURE_DIR = PROJECT_ROOT / 'features_v2_nepal'
monthly = pd.read_parquet(FEATURE_DIR / 'monthly_features.parquet')
daily = pd.read_parquet(FEATURE_DIR / 'daily_anomaly_features.parquet')
shared = pd.read_parquet(FEATURE_DIR / 'shared_features.parquet')

with open(FEATURE_DIR / 'monthly_feature_cols.json') as f:
    monthly_features = json.load(f)
with open(FEATURE_DIR / 'daily_anomaly_feature_cols.json') as f:
    anomaly_features = json.load(f)
with open(FEATURE_DIR / 'shared_feature_cols.json') as f:
    shared_features = json.load(f)

pd.DataFrame([
    {'feature_table': 'monthly', 'rows': len(monthly), 'features': len(monthly_features)},
    {'feature_table': 'daily_anomaly', 'rows': len(daily), 'features': len(anomaly_features)},
    {'feature_table': 'shared', 'rows': len(shared), 'features': len(shared_features)},
])
```

```python
print('Budget risk rate:', round(monthly['target_budget_risk'].mean(), 4))
print('Anomaly rate:', round(daily['target_is_anomaly'].mean(), 4))
print('Late settlement rate:', round(shared['target_late_settlement'].mean(), 4))
```
