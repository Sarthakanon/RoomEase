# 03 - Model Training And Evaluation V2

This notebook presents the final trained model bundle in `models_v2_nepal/`.

## Models Trained

- Monthly spend forecast
- Budget risk classifier
- Daily anomaly detector
- Spending profile clustering
- Shared settlement risk classifier

The best-performing model is the **budget risk classifier** because it has the strongest AUC and F1 score.

```python
from pathlib import Path
import json
import pandas as pd

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks_v2' else Path.cwd()
MODEL_DIR = PROJECT_ROOT / 'models_v2_nepal'
with open(MODEL_DIR / 'training_summary_v2_nepal.json') as f:
    summary = json.load(f)
summary
```

```python
rows = []
for name in ['budget_risk', 'anomaly_detection', 'shared_settlement']:
    item = summary[name]
    rows.append({
        'model': name,
        'accuracy': item.get('accuracy'),
        'precision': item.get('precision'),
        'recall': item.get('recall'),
        'f1': item.get('f1'),
        'auc': item.get('auc'),
        'threshold': item.get('threshold'),
    })
classification_results = pd.DataFrame(rows).sort_values(['auc', 'f1'], ascending=False)
classification_results
```

```python
forecast = summary['monthly_forecast']
improvement = (1 - forecast['mae_npr'] / forecast['baseline_mae_npr']) * 100
print(f"Monthly forecast R2: {forecast['r2']:.3f}")
print(f"Monthly forecast MAE: NPR {forecast['mae_npr']:,.0f}")
print(f"Improvement over baseline MAE: {improvement:.2f}%")
```

```python
best = classification_results.iloc[0]
print('Best performing model:', best['model'])
print(f"AUC: {best['auc']:.3f}")
print(f"F1: {best['f1']:.3f}")
```
